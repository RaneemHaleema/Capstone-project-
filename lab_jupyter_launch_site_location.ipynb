```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Example: Kennedy Space Center coordinates
launch_site_lat = 28.5721
launch_site_lon = -80.6480

# TODO: Use MousePosition on the map to find the closest coastline coordinates
# Example coastline coordinates found by hovering on the map
coastline_lat = 28.56367
coastline_lon = -80.57163

# Function to calculate distance using Haversine formula
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    """
    Calculate distance (km) between two coordinates using Haversine formula.
    """
    R = 6373.0  # approximate radius of Earth in km

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

# Calculate distance to the coastline
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
print(f"Distance from launch site to closest coastline: {distance_coastline:.2f} km")
```

```python
# -------------------------------
# Step 0: Install Folium (only for JupyterLite)
# -------------------------------
import piplite
await piplite.install(['folium'])

# -------------------------------
# Step 1: Import packages
# -------------------------------
import folium
from folium.features import DivIcon
from math import sin, cos, sqrt, atan2, radians

# -------------------------------
# Step 2: Create a base map
# -------------------------------
# Centered around the launch site
launch_site_coord = [28.5721, -80.6480]  # Kennedy Space Center
site_map = folium.Map(location=launch_site_coord, zoom_start=10)

# -------------------------------
# Step 3: Define Haversine function
# -------------------------------
def calculate_distance(lat1, lon1, lat2, lon2):
    R = 6373.0  # approximate radius of Earth in km
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat/2)**2 + cos(lat1)*cos(lat2)*sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    return R * c

# -------------------------------
# Step 4: Nearby point coordinates (e.g., city, railway, highway)
# -------------------------------
nearby_point_coord = [28.5850, -80.6000]  # replace with actual coordinates

# Calculate distance
distance_to_point = calculate_distance(
    launch_site_coord[0], launch_site_coord[1],
    nearby_point_coord[0], nearby_point_coord[1]
)

# -------------------------------
# Step 5: Add marker for the nearby point
# -------------------------------
distance_marker = folium.Marker(
    location=nearby_point_coord,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#FF5733;"><b>{:10.2f} KM</b></div>'.format(distance_to_point)
    )
)
site_map.add_child(distance_marker)

# -------------------------------
# Step 6: Draw a line from launch site to nearby point
# -------------------------------
line_to_point = folium.PolyLine(
    locations=[launch_site_coord, nearby_point_coord],
    color='purple',
    weight=2,
    dash_array='5'
)
site_map.add_child(line_to_point)

# -------------------------------
# Step 7: Display the map
# -------------------------------
site_map
```
